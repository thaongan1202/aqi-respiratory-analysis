```python
import pandas as pd

df = pd.read_csv("AQI_Respiratory_Disease_Final.csv")
```

```python
# 1. BẢNG LOCATION
location_cols = [
    "CountyFIPS",
    "City",
    "County",
    "StateAbbr",
    "State",
    "Latitude",
    "Longitude"
]

location = (
    df[location_cols]
    .drop_duplicates(subset=["CountyFIPS"])
    .sort_values("CountyFIPS")
    .reset_index(drop=True)
)


print(location)
```

```python
# 2. BẢNG AIR QUALITY

air_cols = [
    "CountyFIPS",
    "Year",
    "Days_with_AQI",
    "Good_Days",
    "Moderate_Days",
    "USG_Days",
    "Unhealthy_Days",
    "Very_Unhealthy_Days",
    "Hazardous_Days",
    "Max_AQI",
    "AQI_90th_Percentile",
    "Median_AQI",
    "Days_NO2",
    "Days_Ozone",
    "Days_PM2_5",
    "Days_PM10",
    "AQI_Coverage_Pct",
    "Poor_Air_Days",
    "Poor_Air_Days_Pct",
    "is_extreme_aqi",
    "is_extreme_aqi_high",
    "is_extreme_aqi_low"
]

air_quality = (
    df[air_cols]
    .drop_duplicates(subset=["CountyFIPS", "Year"])
    .sort_values(["CountyFIPS", "Year"])
    .reset_index(drop=True)
)

print(air_quality)
```

```python
# 4. BẢNG RESPIRATORY DISEASE
# Mục đích:
# - Phân tích bệnh hô hấp
# - So sánh COPD và Current Asthma
# - So sánh tỷ lệ bệnh với AQI
#
# PK: CountyFIPS + Year + Disease
# FK: CountyFIPS -> Location

disease_cols = [
    "CountyFIPS",
    "Year",
    "Disease",
    "Disease_AdjPrev",
    "is_extreme_disease"
]

respiratory_disease = (
    df[disease_cols]
    .drop_duplicates(
        subset=["CountyFIPS", "Year", "Disease"]
    )
    .sort_values(
        ["CountyFIPS", "Year", "Disease"]
    )
    .reset_index(drop=True)
)

print(respiratory_disease)
```

```python
# 5. BẢNG DEMOGRAPHIC
# Mục đích:
# - Chứa các yếu tố nền
# - Population
# - Smoking
#
# PK: CountyFIPS + Year
# FK: CountyFIPS -> Location

demographic_cols = [
    "CountyFIPS",
    "Year",

    "Population",
    "Log_Population",

    "Smoking_AdjPrev",
    "Complete_4_Years",

    "is_extreme_smoking",
    "is_extreme_population"
]

demographic = (
    df[demographic_cols]
    .drop_duplicates(subset=["CountyFIPS", "Year"])
    .sort_values(["CountyFIPS", "Year"])
    .reset_index(drop=True)
)

print(demographic)
```

```python
#6. KIỂM TRA KHÓA CHÍNH CÓ BỊ TRÙNG HAY KHÔNG
# Location
print(
    "Location:",
    location.duplicated(
        subset=["CountyFIPS"]
    ).sum()
)

# Air Quality
print(
    "AirQuality:",
    air_quality.duplicated(
        subset=["CountyFIPS", "Year"]
    ).sum()
)

# Respiratory Disease
print(
    "RespiratoryDisease:",
    respiratory_disease.duplicated(
        subset=["CountyFIPS", "Year", "Disease"]
    ).sum()
)

# Demographic
print(
    "Demographic:",
    demographic.duplicated(
        subset=["CountyFIPS", "Year"]
    ).sum()
)
```

```python
# XUẤT CSV
location.to_csv("Location.csv", index=False, encoding="utf-8-sig")
air_quality.to_csv("AirQuality.csv", index=False, encoding="utf-8-sig")
respiratory_disease.to_csv("RespiratoryDisease.csv", index=False, encoding="utf-8-sig")
demographic.to_csv("Demographic.csv", index=False, encoding="utf-8-sig")
```
